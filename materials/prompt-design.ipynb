Prompt design is the **most important factor** in agent frameworks like CrewAI.

Here’s a clear, battle-tested set of **rules for designing prompts in Agent Frameworks** (especially CrewAI):

---

### 1. Core Principles (Golden Rules)

| Rule | Why It Matters | How to Apply |
|------|----------------|--------------|
| **Be Extremely Specific** | Vague prompts = hallucination & tool skipping | Use step-by-step workflows, exact output formats |
| **Role + Goal + Constraints** | Gives the model identity and boundaries | Always include `role`, `goal`, `backstory` |
| **Force Tool Use** | Models love to shortcut | Explicitly say “You MUST use tools”, “Never answer from memory” |
| **Few-Shot Examples** | Best way to teach behavior | Show 1–2 examples of Thought → Tool Call → Observation |
| **Output Format Control** | Prevents rambling | Define exact final output structure |

### 3. Key Prompt Design Rules (Practical Guide)

**1. Task Description (Most Important)**
- Break into **numbered steps** (1., 2., 3.)
- Use words like: `MUST`, `STRICTLY`, `EXACTLY`, `NEVER`
- Include success criteria

**2. Backstory**
- Give strong identity
- Include negative instructions ("Never...", "Do not...")
- Add consequences ("You will be penalized if...")

**3. Goal**
- Make it narrow and actionable

**4. Handle Tool Resistance**
Add this to every tool-heavy agent:

```python
"You are a tool-using machine. You NEVER answer from internal knowledge when tools are available."
```

**5. Control Final Output**
Always define the exact format you want:

```python
"Finally, output ONLY in this exact format:
**Alpha Name:** ...
**Economic Hypothesis:** ..."
```

### 5. General Tips for CrewAI

- Keep `temperature=0.1` for tool-heavy agents (`researcher`, `coder`, `validator`)
- Keep `temperature=0.3~0.4` for creative agents (`ideator`)
- Use `verbose=True` during development
- Test one agent at a time before running full crew
- Add more negative instructions than positive ones
